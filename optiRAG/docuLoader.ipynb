#### 文档加载器

```python
from langchain_community.document_loaders import (
    PyPDFLoader,
    TextLoader,
    Docx2txtLoader,
    DirectoryLoader, # 批量加载
    CSVLoader
)
```

```python
# 加载pdf
from pathlib import Path

# base_dir = Path.cwd()  # 或者你自己指定到 optiRAG 目录
# pdf_path = (base_dir / "02-load.pdf").resolve()
# print(pdf_path, pdf_path.exists())
pdf_loader = PyPDFLoader("../02-load.pdf")
pdf_docs = pdf_loader.load()

print(f"加载了 {len(pdf_docs)} documents")
print("第一页的内容预览：")
print(pdf_docs[0].page_content[:100])
print("Documents结构：")
print(pdf_docs)
```

```python
# 加载word
docx_loader = Docx2txtLoader("../01-load.docx")
docx_docs = docx_loader.load()

print(f"加载了 {len(docx_docs)} documents")
print("第一页的内容预览：")
print(docx_docs[0].page_content[:100])
```

```python
# 加载txt
txt_loader = TextLoader("../test.txt",encoding="utf-8")
txt_docs = txt_loader.load()
print(f"加载了 {len(txt_docs)} documents")
print("第一页的内容预览：")
print(txt_docs[0].page_content[:100])
```

```python
from langchain_community.document_loaders import BSHTMLLoader

html_path = "../05-load.html"

# Windows 下默认编码常是 gbk，这里显式指定以避免 UnicodeDecodeError
try:
    loader = BSHTMLLoader(html_path, open_encoding="utf-8")
    docs = loader.load()
except UnicodeDecodeError:
    # 若文件实际是国标编码，回退到 gb18030（兼容 gbk）
    loader = BSHTMLLoader(html_path, open_encoding="gb18030")
    docs = loader.load()

print(docs[0].page_content[:20])

# 如果你想保留语义切分，可用 UnstructuredHTMLLoader；但它可能依赖额外 NLTK 资源
```

```python
# 加载目录下的所有文件
dir_loader = DirectoryLoader(
    "../",
    glob = "**/*.pdf", # 匹配所有pdf文件
    loader_cls = PyPDFLoader
)
all_docs = dir_loader.load()
# print(all_docs)
print(f"加载了 {len(all_docs)} documents")
print("第一页的内容预览：")
print(all_docs[0].page_content[:100])
```
